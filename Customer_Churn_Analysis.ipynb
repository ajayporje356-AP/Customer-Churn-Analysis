This confirms that the above customers have just signed up, so their time with the company (tenure) is exactly **0 months**, and they haven't paid their first bill yet.


During our data inspection, we found 11 missing values in the TotalCharges column. By investigating further, we discovered that all 11 of these customers have a tenure of 0. This means they are brand new customers who have just signed up and haven't been billed yet.

Since our main goal is to understand customer churn (why people leave), brand new customers don't give us any historical data to learn from. Furthermore, 11 rows represent a tiny fraction (about 0.15%) of our 7,043 total records. Because of this, the most logical step is to drop these 11 rows from our analysis so they don't skew our results.

```python
df.dropna(inplace=True)
df.shape
```

Calculating the Churn Rate

While raw counts are helpful for understanding volume, business stakeholders track success using percentages. The **Churn Rate** represents the proportion of our total customer base that has left.


```python
pd.crosstab(df["Contract"],df["Churn"])
```

This confirms that the Month-to-Month contracts have a higher likelihood of churning compared to those on 1-year or 2-year commitments.

```python
# Checking avg tenure of churning customers
df.groupby("Churn")["tenure"].mean()
```

We found a significant difference in customer lifespan. Customers who stay have an average tenure of roughly 38 months, whereas customers who churn leave after an average of just 18 months. This indicates that the first year and a half is the highest-risk period for customer attrition.

```python
# Analyzing the Impact of Price
df.groupby("Churn")["MonthlyCharges"].mean()
```

We investigated whether price impacts customer retention. The data shows that customers who churn have a significantly higher average monthly bill compared to those who stay. This suggests that price sensitivity is a major factor in attrition.
